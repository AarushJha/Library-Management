# Library Management

## Module 1: Add books

```python
books = {
    '101': ['Concepts of Physics (Vol 1)', 'H.C. Verma', 'Available'],
    '102': ['Problems in General Physi|s', 'I.E. Irodov', 'Issued'],
    '103': ['Physics Galaxy', 'Ashish Arora', 'Issued'],
    '104': ['Understanding Physics Series', 'D.C. Pandey', 'Available'],
    '105': ['Problems in Physical Chemistry', 'N. Awasthi', 'Issued'],
    '106': ['Advanced Problems in Organic Chemistry', 'M.S. Chauhan', 'Available'],
    '107': ['Problems in Inorganic Chemistry', 'V.K. Jaiswal', 'Issued'],
    '108': ['Cengage Mathematics Series', 'G. Tewani', 'Available'],
    '109': ['Advanced Problems in Mathematics', 'Vikas Gupta', 'Available'],
    '110': ['Concise Inorganic Chemistry', 'J.D. Lee', 'Issued']
}
```

```python
def add_books(data_dict):
    book_id = input("Enter book I.D.: ")
    book_name = input("Enter name of the book: ")
    book_author = input("Enter author of the book: ")
    book_status = input("Is book issued? [Yes/No] ")
    book_issue_status = ''
    
    if book_status == "Yes":
        book_issue_status = "Issued"
    else:
        book_issue_status = "Available"

    data_dict[book_id] = [book_name, book_author, book_issue_status]
```

```python
add_books(books)
```

```python
print(books)
```

## Module 2: Display Books

```python
import pandas as pd
```

```python
books_df = pd.DataFrame(books).T
books_df.columns = ['Book Name', 'Author', 'Issue Status']
print(books_df)
```

## Module 3: Search for a Book

```python
def search_book(df):
    book_id_df = input("Enter Book I.D.: ")

    if book_id_df in df.index:
        book = df.loc[book_id_df]
        print(f"Book Name: {book['Book Name']}")
        print(f"Author: {book['Author']}")
        print(f"Issue Status: {book['Issue Status']}")
    else:
        print(f"No book found with I.D. {book_id_df}")
```

```python
search_book(books_df)
```

## Module 4: Issue/Return a Book

```python
def issue_return_book(df):
    x = input("Press [1] for Issue and [2] for Return: ")
    if x == "1":
        book_id = input("Enter Book I.D.: ")
    
        if book_id not in df.index:
            print(f"No book found with I.D. {book_id}")
        elif df.loc[book_id, 'Issue Status'] == 'Issued':
            print("Sorry, this book is already issued.")
        else:
            df.loc[book_id, 'Issue Status'] = 'Issued'
            print("Book issued successfully.")
    elif x == "2":
        book_id = input("Enter Book I.D.: ")

        if book_id not in df.index:
            print(f"No book found with I.D. {book_id}")
        elif df.loc[book_id, 'Issue Status'] == 'Available':
            print("Sorry, this book cannot be returned")
        else:
            df.loc[book_id, 'Issue Status'] = 'Available'
            print("Book returned successfully.")
    else:
        print("Invalid Choice, press [1] or [2]")
```

```python
issue_return_book(books_df)
```
